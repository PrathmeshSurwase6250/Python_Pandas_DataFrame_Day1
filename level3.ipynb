21. Find students whose marks are greater than 80.

```python
students[students['marks'] > 80]
```

22. Find students whose marks are below 70.

```python
students[students['marks'] < 70]
```

23. Find students from Pune.

```python
students[students["city"] =='Pune']
```

24. Find students enrolled in Python.

```python
students[students['course'] == "Python"]
```

25. Find students from Pune who scored above 80.

```python
students[(students['city'] == 'Pune') & (students['marks'] > 80)]
```

26. Find students whose marks are between 70 and 90.

```python
students[students['marks'].between(70,90)]
```

27. Find students whose attendance is greater than 90.

```python
students[students['attendance'] > 90]
```

28. Find female students from Mumbai.

```python
students[(students['city'] == 'Mumbai') & (students['gender'] == 'F')]
```

29. Find students who paid more than ₹50,000.

```python
students[students['fees_paid'] > 50000]
```

30. Find students whose marks are either below 60 or above 90.

```python
students[(students['marks'] < 60) | (students["marks"] > 90)]
```
